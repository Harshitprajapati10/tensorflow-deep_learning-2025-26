*3) Create two tensors containing random values between 0 and 1 with shape [5, 300].*

```python
#create two different random tensors

random_1 = tf.random.Generator.from_seed(42)
random_1 = random_1.normal(shape=(5,300))

random_2 = tf.random.Generator.from_seed(91)
random_2 = random_2.normal(shape=(5,300))
random_1, random_2
```

*4) Multiply the above two tensors*

```python
random_2_T = tf.transpose(random_2)
C = tf.matmul(random_1, random_2_T)
C
```

*5) multiply the above two tensors using dot product*

```python
#dot product is same as matmul

dot_product = tf.tensordot(random_1, random_2_T, axes=1)
dot_product
```

*6) Create a tensor with random values between 0 and 1 with shape [224, 224, 3]*

```python
random_3 = tf.random.Generator.from_seed(42)
random_3 = random_3.normal(shape=(224,224,3))
random_3
```

7) Find the min and max values of the tensor you created in 6.

```python
min_val = tf.reduce_min(random_3).numpy()
max_val = tf.reduce_max(random_3).numpy()
min_val, max_val
```

9) Create a tensor with shape [10] using your own choice of values, then find the index which has the maximum value

```python
import numpy as np
arr = np.arange(10)
random_5 = tf.constant(arr)
random_5, random_5.shape
```

```python
min_index = tf.argmin(random_5).numpy()
max_index = tf.argmax(random_5).numpy()
min_index, max_index
```
